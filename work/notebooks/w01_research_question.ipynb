## 2. The question: decision, action, cost of a wrong call

Decision this improves: which pages a content reviewer looks at first, out of a large
inventory, when review time is limited.

Who acts on it: a content/SEO team member with a fixed weekly review capacity
(e.g. can only review ~50 pages a week).

Cost of a wrong call:
- False positive (flagged as needing review, but it wasn't actually declining) ->
  wasted reviewer time on a page that didn't need attention.
- False negative (a genuinely declining page never surfaces in the top-K) ->
  the page keeps losing traffic/visibility undetected, which compounds over time.
Both costs matter, but a missed real decline is more expensive long-term than a
wasted review, since traffic loss is harder to reverse the longer it's ignored.
